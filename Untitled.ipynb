# Prompt
I am working with an online retail dataset that just been loaded ysubg pandas. I need to create a comprehensive data quality assessment function that can :
count and calculate the percentage of missing values for each column 
detect duplicates records
identify outliers in numeri columns using IQR method
Provide a summary report of all findings

```python

import pandas as pd
import numpy as np


def data_quality_assessment(df):
    """
    Comprehensive Data Quality Assessment

    Checks:
    1. Missing values (count and percentage)
    2. Duplicate records
    3. Numerical outliers using IQR method
    4. Overall data quality summary

    Returns:
    - missing_report
    - duplicate_report
    - outlier_report
    - summary_report
    """

    # Validate input
    if not isinstance(df, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    if df.empty:
        raise ValueError("DataFrame is empty.")

    total_rows = len(df)
    total_columns = len(df.columns)

    # --------------------------------
    # 1. MISSING VALUES ASSESSMENT
    # --------------------------------

    missing_count = df.isnull().sum()

    missing_percentage = (
        missing_count / total_rows
    ) * 100

    missing_report = pd.DataFrame({
        'Missing_Count': missing_count,
        'Missing_Percentage': missing_percentage.round(2)
    })

    missing_report = missing_report.sort_values(
        by='Missing_Percentage',
        ascending=False
    )

    # --------------------------------
    # 2. DUPLICATE RECORDS ASSESSMENT
    # --------------------------------

    duplicate_count = df.duplicated().sum()

    duplicate_percentage = (
        duplicate_count / total_rows
    ) * 100

    duplicate_report = pd.DataFrame({
        'Metric': [
            'Total Records',
            'Duplicate Records',
            'Unique Records',
            'Duplicate Percentage'
        ],
        'Value': [
            total_rows,
            duplicate_count,
            total_rows - duplicate_count,
            round(duplicate_percentage, 2)
        ]
    })

    # --------------------------------
    # 3. OUTLIER DETECTION USING IQR
    # --------------------------------

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns

    outlier_results = []

    for column in numerical_columns:

        series = df[column].dropna()

        if series.empty:
            continue

        # Calculate quartiles
        Q1 = series.quantile(0.25)
        Q3 = series.quantile(0.75)

        # Calculate IQR
        IQR = Q3 - Q1

        # Calculate boundaries
        lower_bound = Q1 - (1.5 * IQR)
        upper_bound = Q3 + (1.5 * IQR)

        # Identify outliers
        outliers = series[
            (series < lower_bound) |
            (series > upper_bound)
        ]

        outlier_count = len(outliers)

        outlier_percentage = (
            outlier_count / len(series)
        ) * 100

        outlier_results.append({
            'Column': column,
            'Q1': round(Q1, 2),
            'Q3': round(Q3, 2),
            'IQR': round(IQR, 2),
            'Lower_Bound': round(lower_bound, 2),
            'Upper_Bound': round(upper_bound, 2),
            'Outlier_Count': outlier_count,
            'Outlier_Percentage': round(
                outlier_percentage, 2
            )
        })

    outlier_report = pd.DataFrame(outlier_results)

    # --------------------------------
    # 4. OVERALL DATA QUALITY SUMMARY
    # --------------------------------

    total_missing = int(missing_count.sum())

    columns_with_missing = int(
        (missing_count > 0).sum()
    )

    total_outliers = (
        int(outlier_report['Outlier_Count'].sum())
        if not outlier_report.empty else 0
    )

    summary_report = pd.DataFrame({
        'Metric': [
            'Total Rows',
            'Total Columns',
            'Total Missing Values',
            'Columns With Missing Values',
            'Total Duplicate Records',
            'Duplicate Percentage',
            'Numerical Columns Assessed',
            'Total Outliers Detected'
        ],

        'Value': [
            total_rows,
            total_columns,
            total_missing,
            columns_with_missing,
            duplicate_count,
            round(duplicate_percentage, 2),
            len(numerical_columns),
            total_outliers
        ]
    })

    # --------------------------------
    # 5. DISPLAY RESULTS
    # --------------------------------

    print("=" * 55)
    print("       DATA QUALITY ASSESSMENT REPORT")
    print("=" * 55)

    print("\n1. MISSING VALUES REPORT")
    print("-" * 55)
    print(missing_report.to_string())

    print("\n2. DUPLICATE RECORDS REPORT")
    print("-" * 55)
    print(duplicate_report.to_string(index=False))

    print("\n3. NUMERICAL OUTLIER REPORT")
    print("-" * 55)

    if not outlier_report.empty:
        print(outlier_report.to_string(index=False))
    else:
        print("No numerical columns available.")

    print("\n4. OVERALL DATA QUALITY SUMMARY")
    print("-" * 55)
    print(summary_report.to_string(index=False))

    print("\n" + "=" * 55)
    print("       ASSESSMENT COMPLETED SUCCESSFULLY")
    print("=" * 55)

    # Return all reports
    return (
        missing_report,
        duplicate_report,
        outlier_report,
        summary_report
    )


# ==========================================
# EXECUTE THE DATA QUALITY ASSESSMENT
# ==========================================

# Assuming your dataset is already loaded as df

missing_report, duplicate_report, outlier_report, summary_report = (
    data_quality_assessment(df)
)
```

Can you help me create an intelligent imputation function that:
1. Automatically detects the distribution of numeric columns
2. Chooses mean for notami distribution, median for skewed distribution
3. Uses mode for categorical columns.
4. Provide before/after comparisions.
5. Compare quality metrics before and after


```python

# ================================================================
# INTELLIGENT DATA IMPUTATION PIPELINE
# PROJECT: ONLINE RETAIL DATASET
# ================================================================

import pandas as pd
import numpy as np


# ================================================================
# SECTION 1: CONSTANT CONFIGURATION
# ALL USER-ADJUSTABLE INPUTS ARE HERE
# ================================================================

# Input dataset (already loaded in Jupyter Notebook)
INPUT_DF = df

# Statistical configuration
SKEW_THRESHOLD = 0.5
IQR_MULTIPLIER = 1.5

# Special category for missing CustomerID
CUSTOMER_ID_COLUMN = 'CustomerID'
CUSTOMER_ID_SENTINEL = -1

# Identifier columns: excluded from statistical imputation
IDENTIFIER_COLUMNS = [
    'InvoiceNo',
    'StockCode',
    'CustomerID'
]

# Date columns: excluded from automatic imputation
DATE_COLUMNS = [
    'InvoiceDate'
]

# Product description configuration
DESCRIPTION_COLUMN = 'Description'
PRODUCT_CODE_COLUMN = 'StockCode'

# Optional fallback for unmatched descriptions
DESCRIPTION_MODE_FALLBACK = False

# Distribution report configuration
TOP_N_CATEGORIES = 5

# Excel export configuration
EXPORT_REPORTS = True
EXPORT_FILE = 'Online_Retail_Imputation_Report.xlsx'


# ================================================================
# SECTION 2: HELPER FUNCTIONS
# ================================================================

def calculate_outliers(data, columns):

    results = {}

    for col in columns:

        values = data[col].dropna()

        if len(values) < 2:
            results[col] = 0
            continue

        q1 = values.quantile(0.25)
        q3 = values.quantile(0.75)

        iqr = q3 - q1

        lower_bound = q1 - IQR_MULTIPLIER * iqr
        upper_bound = q3 + IQR_MULTIPLIER * iqr

        outlier_count = (
            (values < lower_bound) |
            (values > upper_bound)
        ).sum()

        results[col] = int(outlier_count)

    return results


def calculate_quality_metrics(data, numerical_columns):

    total_cells = data.size

    total_missing = int(data.isnull().sum().sum())

    missing_percentage = (
        total_missing / total_cells * 100
        if total_cells > 0 else 0
    )

    outliers = calculate_outliers(
        data,
        numerical_columns
    )

    return {
        'Total Rows': len(data),
        'Total Columns': len(data.columns),
        'Total Missing Values': total_missing,
        'Missing Percentage': round(
            missing_percentage, 2
        ),
        'Duplicate Records': int(
            data.duplicated().sum()
        ),
        'Total Numerical Outliers': sum(
            outliers.values()
        )
    }


def get_top_categories(series, stage):

    distribution = (
        series.value_counts(dropna=False)
        .head(TOP_N_CATEGORIES)
    )

    return pd.DataFrame({
        'Stage': stage,
        'Value': distribution.index.astype(str),
        'Frequency': distribution.values
    })


# ================================================================
# SECTION 3: MAIN INTELLIGENT IMPUTATION FUNCTION
# ================================================================

def intelligent_imputation(data):

    # Validate input
    if not isinstance(data, pd.DataFrame):
        raise TypeError(
            "Input must be a pandas DataFrame."
        )

    if data.empty:
        raise ValueError(
            "Input DataFrame is empty."
        )

    # Work on a copy to preserve original dataset
    cleaned_df = data.copy(deep=True)

    # Check configured columns
    if (
        CUSTOMER_ID_COLUMN in cleaned_df.columns
        and CUSTOMER_ID_SENTINEL is not None
    ):
        pass

    # ------------------------------------------------------------
    # A. IDENTIFY ELIGIBLE NUMERICAL COLUMNS
    # ------------------------------------------------------------

    numerical_columns = [
        col for col in cleaned_df.select_dtypes(
            include=np.number
        ).columns
        if col not in IDENTIFIER_COLUMNS
        and col not in DATE_COLUMNS
    ]

    # ------------------------------------------------------------
    # B. IDENTIFY ELIGIBLE CATEGORICAL COLUMNS
    # ------------------------------------------------------------

    categorical_columns = [
        col for col in cleaned_df.select_dtypes(
            include=['object', 'category', 'string']
        ).columns
        if col not in IDENTIFIER_COLUMNS
        and col not in DATE_COLUMNS
        and col != DESCRIPTION_COLUMN
    ]

    # ------------------------------------------------------------
    # C. CAPTURE BEFORE-IMPUTATION METRICS
    # ------------------------------------------------------------

    before_metrics = calculate_quality_metrics(
        cleaned_df,
        numerical_columns
    )

    before_missing = cleaned_df.isnull().sum()

    before_outliers = calculate_outliers(
        cleaned_df,
        numerical_columns
    )

    strategy_results = []

    # ------------------------------------------------------------
    # D. CUSTOMER ID SPECIAL CATEGORY IMPUTATION
    # ------------------------------------------------------------

    customer_distribution = []

    if CUSTOMER_ID_COLUMN in cleaned_df.columns:

        original_customer = cleaned_df[
            CUSTOMER_ID_COLUMN
        ].copy()

        missing_count = original_customer.isnull().sum()

        # Capture distribution before imputation
        customer_distribution.append(
            get_top_categories(
                original_customer,
                'Before Imputation'
            )
        )

        if missing_count > 0:

            # Replace missing CustomerIDs with -1
            cleaned_df[CUSTOMER_ID_COLUMN] = (
                cleaned_df[CUSTOMER_ID_COLUMN].fillna(
                    CUSTOMER_ID_SENTINEL
                )
            )

            # Convert to nullable integer if all values
            # are integer-like
            numeric_ids = pd.to_numeric(
                cleaned_df[CUSTOMER_ID_COLUMN],
                errors='coerce'
            )

            if (
                numeric_ids.notna().all()
                and np.allclose(
                    numeric_ids,
                    np.round(numeric_ids)
                )
            ):
                cleaned_df[CUSTOMER_ID_COLUMN] = (
                    numeric_ids.astype('int64')
                )

        strategy_results.append({
            'Column': CUSTOMER_ID_COLUMN,
            'Data_Type': 'Identifier',
            'Skewness': np.nan,
            'Strategy': 'Special Category (-1)',
            'Missing_Before': int(missing_count),
            'Imputation_Value': CUSTOMER_ID_SENTINEL,
            'Missing_After': int(
                cleaned_df[CUSTOMER_ID_COLUMN].isnull().sum()
            )
        })

        # Capture distribution after imputation
        customer_distribution.append(
            get_top_categories(
                cleaned_df[CUSTOMER_ID_COLUMN],
                'After Imputation'
            )
        )

    # ------------------------------------------------------------
    # E. NUMERICAL IMPUTATION
    # Mean for approximately symmetric distributions
    # Median for skewed distributions
    # ------------------------------------------------------------

    for col in numerical_columns:

        missing_count = cleaned_df[col].isnull().sum()

        skewness = cleaned_df[col].skew()

        if pd.isna(skewness):
            strategy = 'Median'

        elif abs(skewness) <= SKEW_THRESHOLD:
            strategy = 'Mean'

        else:
            strategy = 'Median'

        fill_value = np.nan

        if missing_count > 0:

            if strategy == 'Mean':
                fill_value = cleaned_df[col].mean()
            else:
                fill_value = cleaned_df[col].median()

            if pd.notna(fill_value):
                cleaned_df[col] = cleaned_df[col].fillna(
                    fill_value
                )

        strategy_results.append({
            'Column': col,
            'Data_Type': 'Numerical',
            'Skewness': round(skewness, 3)
                if pd.notna(skewness) else np.nan,
            'Strategy': strategy,
            'Missing_Before': int(missing_count),
            'Imputation_Value': fill_value,
            'Missing_After': int(
                cleaned_df[col].isnull().sum()
            )
        })

    # ------------------------------------------------------------
    # F. DESCRIPTION IMPUTATION USING STOCKCODE
    # ------------------------------------------------------------

    if (
        DESCRIPTION_COLUMN in cleaned_df.columns
        and PRODUCT_CODE_COLUMN in cleaned_df.columns
    ):

        missing_count = cleaned_df[
            DESCRIPTION_COLUMN
        ].isnull().sum()

        inferred_count = 0
        fallback_count = 0

        if missing_count > 0:

            valid_pairs = cleaned_df[
                [PRODUCT_CODE_COLUMN, DESCRIPTION_COLUMN]
            ].dropna()

            # Identify product codes having one consistent
            # description
            description_counts = (
                valid_pairs.groupby(PRODUCT_CODE_COLUMN)[
                    DESCRIPTION_COLUMN
                ].nunique()
            )

            consistent_codes = description_counts[
                description_counts == 1
            ].index

            mapping = (
                valid_pairs[
                    valid_pairs[PRODUCT_CODE_COLUMN].isin(
                        consistent_codes
                    )
                ]
                .drop_duplicates(PRODUCT_CODE_COLUMN)
                .set_index(PRODUCT_CODE_COLUMN)[
                    DESCRIPTION_COLUMN
                ]
                .to_dict()
            )

            missing_mask = cleaned_df[
                DESCRIPTION_COLUMN
            ].isnull()

            inferred = cleaned_df.loc[
                missing_mask, PRODUCT_CODE_COLUMN
            ].map(mapping)

            inferred_count = int(
                inferred.notna().sum()
            )

            cleaned_df.loc[
                missing_mask, DESCRIPTION_COLUMN
            ] = inferred

            # Optional mode fallback
            if DESCRIPTION_MODE_FALLBACK:

                remaining_mask = cleaned_df[
                    DESCRIPTION_COLUMN
                ].isnull()

                modes = cleaned_df[
                    DESCRIPTION_COLUMN
                ].mode()

                if not modes.empty:

                    fallback_count = int(
                        remaining_mask.sum()
                    )

                    cleaned_df.loc[
                        remaining_mask,
                        DESCRIPTION_COLUMN
                    ] = modes.iloc[0]

        strategy_results.append({
            'Column': DESCRIPTION_COLUMN,
            'Data_Type': 'Categorical',
            'Skewness': np.nan,
            'Strategy': 'StockCode Mapping',
            'Missing_Before': int(missing_count),
            'Imputation_Value': (
                f'{inferred_count} inferred, '
                f'{fallback_count} mode fallback'
            ),
            'Missing_After': int(
                cleaned_df[
                    DESCRIPTION_COLUMN
                ].isnull().sum()
            )
        })

    # ------------------------------------------------------------
    # G. MODE IMPUTATION FOR OTHER CATEGORICAL COLUMNS
    # ------------------------------------------------------------

    for col in categorical_columns:

        missing_count = cleaned_df[col].isnull().sum()

        if missing_count == 0:
            continue

        modes = cleaned_df[col].mode(
            dropna=True
        )

        if modes.empty:
            continue

        fill_value = modes.iloc[0]

        cleaned_df[col] = cleaned_df[col].fillna(
            fill_value
        )

        strategy_results.append({
            'Column': col,
            'Data_Type': 'Categorical',
            'Skewness': np.nan,
            'Strategy': 'Mode',
            'Missing_Before': int(missing_count),
            'Imputation_Value': fill_value,
            'Missing_After': int(
                cleaned_df[col].isnull().sum()
            )
        })

    # ------------------------------------------------------------
    # H. CAPTURE AFTER-IMPUTATION METRICS
    # ------------------------------------------------------------

    after_metrics = calculate_quality_metrics(
        cleaned_df,
        numerical_columns
    )

    after_missing = cleaned_df.isnull().sum()

    after_outliers = calculate_outliers(
        cleaned_df,
        numerical_columns
    )

    # ------------------------------------------------------------
    # I. IMPUTATION STRATEGY REPORT
    # ------------------------------------------------------------

    strategy_report = pd.DataFrame(
        strategy_results
    )

    # ------------------------------------------------------------
    # J. MISSING VALUE COMPARISON
    # ------------------------------------------------------------

    missing_comparison = pd.DataFrame({

        'Missing_Before': before_missing,

        'Missing_After': after_missing,

        'Values_Imputed': (
            before_missing - after_missing
        ),

        'Percentage_Before': (
            before_missing / len(data) * 100
        ).round(2),

        'Percentage_After': (
            after_missing / len(data) * 100
        ).round(2)

    })

    # ------------------------------------------------------------
    # K. OVERALL QUALITY COMPARISON
    # ------------------------------------------------------------

    comparison_report = pd.DataFrame({

        'Metric': list(before_metrics.keys()),

        'Before_Imputation': list(
            before_metrics.values()
        ),

        'After_Imputation': list(
            after_metrics.values()
        )

    })

    # ------------------------------------------------------------
    # L. OUTLIER COMPARISON
    # ------------------------------------------------------------

    outlier_comparison = pd.DataFrame({

        'Column': numerical_columns,

        'Outliers_Before': [
            before_outliers[col]
            for col in numerical_columns
        ],

        'Outliers_After': [
            after_outliers[col]
            for col in numerical_columns
        ]

    })

    # ------------------------------------------------------------
    # M. CUSTOMER ID DISTRIBUTION COMPARISON
    # ------------------------------------------------------------

    if customer_distribution:

        customer_distribution_report = pd.concat(
            customer_distribution,
            ignore_index=True
        )

    else:

        customer_distribution_report = pd.DataFrame(
            columns=['Stage', 'Value', 'Frequency']
        )

    # ------------------------------------------------------------
    # N. DISPLAY ALL REPORTS
    # ------------------------------------------------------------

    print("\n" + "=" * 70)
    print("       INTELLIGENT DATA IMPUTATION REPORT")
    print("=" * 70)

    print("\n1. IMPUTATION STRATEGIES")
    print("-" * 70)
    print(strategy_report.to_string(index=False))

    print("\n2. MISSING VALUES BEFORE AND AFTER")
    print("-" * 70)
    print(missing_comparison.to_string())

    print("\n3. OVERALL DATA QUALITY COMPARISON")
    print("-" * 70)
    print(comparison_report.to_string(index=False))

    print("\n4. NUMERICAL OUTLIER COMPARISON")
    print("-" * 70)
    print(outlier_comparison.to_string(index=False))

    print("\n5. CUSTOMER ID DISTRIBUTION")
    print("-" * 70)
    print(customer_distribution_report.to_string(index=False))

    print("\n" + "=" * 70)
    print("       IMPUTATION COMPLETED SUCCESSFULLY")
    print("=" * 70)

    return (
        cleaned_df,
        strategy_report,
        comparison_report,
        missing_comparison,
        outlier_comparison,
        customer_distribution_report
    )


# ================================================================
# SECTION 4: EXECUTE THE IMPUTATION PIPELINE
# ================================================================

(
    cleaned_df,
    strategy_report,
    comparison_report,
    missing_comparison,
    outlier_comparison,
    customer_distribution_report
) = intelligent_imputation(INPUT_DF)


# ================================================================
# SECTION 5: EXPORT ALL RESULTS TO EXCEL
# ================================================================

if EXPORT_REPORTS:

    with pd.ExcelWriter(
        EXPORT_FILE,
        engine='openpyxl'
    ) as writer:

        cleaned_df.to_excel(
            writer,
            sheet_name='Cleaned Dataset',
            index=False
        )

        strategy_report.to_excel(
            writer,
            sheet_name='Imputation Strategies',
            index=False
        )

        comparison_report.to_excel(
            writer,
            sheet_name='Quality Comparison',
            index=False
        )

        missing_comparison.to_excel(
            writer,
            sheet_name='Missing Comparison'
        )

        outlier_comparison.to_excel(
            writer,
            sheet_name='Outlier Comparison',
            index=False
        )

        customer_distribution_report.to_excel(
            writer,
            sheet_name='Customer Distribution',
            index=False
        )

    print("\nExcel report exported:", EXPORT_FILE)


# ================================================================
# SECTION 6: FINAL DATASET VALIDATION
# ================================================================

print("\nFINAL DATASET VALIDATION")
print("-" * 70)

print("Original dataset shape:", INPUT_DF.shape)
print("Cleaned dataset shape:", cleaned_df.shape)

print("\nRemaining missing values:")
print(cleaned_df.isnull().sum())

print("\nCustomerID data type:")
print(cleaned_df[CUSTOMER_ID_COLUMN].dtype)

print("\nCleaned dataset preview:")
display(cleaned_df.head())
```

Next, I need to handle duplicate records in my retail dataset. Can you help me create a function that:

1. Identifies exact duplicates across all columns.
2. Provides statistics on duplicate patterns
3. Show a few duplicate records
4. Removes the duplicates while preserving the first occurence
5. Reports the number of records removed
6. Validates that duplicates have been successfully removed



We need to ensure that the duplicates are meant to be or not, for example in customer id, or description, country categorical variables, and once which are supposed to not have duplicates like the invoice no, stock no etc

```python

# ================================================================
# DUPLICATE DETECTION AND REMOVAL PIPELINE
# PROJECT: ONLINE RETAIL DATASET
# ================================================================

import pandas as pd
import numpy as np


# ================================================================
# SECTION 1: CONSTANT CONFIGURATION
# ALL USER-ADJUSTABLE INPUTS ARE HERE
# ================================================================

# Input dataset
INPUT_DF = cleaned_df

# Duplicate removal configuration
KEEP_FIRST_OCCURRENCE = True

# Number of duplicate records to display
SAMPLE_SIZE = 10

# Number of duplicate patterns to display
TOP_DUPLICATE_PATTERNS = 10

# Business-key combinations for investigation only
# These are NOT automatically removed.
REVIEW_SUBSETS = {
    'InvoiceNo': ['InvoiceNo'],
    'StockCode': ['StockCode'],
    'CustomerID': ['CustomerID'],
    'InvoiceNo + StockCode': [
        'InvoiceNo', 'StockCode'
    ]
}

# Excel export configuration
EXPORT_REPORTS = True
EXPORT_FILE = 'Online_Retail_Duplicate_Report.xlsx'


# ================================================================
# SECTION 2: MAIN DUPLICATE DETECTION FUNCTION
# ================================================================

def handle_duplicates(data):

    # ------------------------------------------------------------
    # A. VALIDATE INPUT
    # ------------------------------------------------------------

    if not isinstance(data, pd.DataFrame):
        raise TypeError(
            "Input must be a pandas DataFrame."
        )

    if data.empty:
        raise ValueError(
            "Input DataFrame is empty."
        )

    original_df = data.copy(deep=True)

    total_rows_before = len(original_df)

    total_columns = len(original_df.columns)

    # ------------------------------------------------------------
    # B. IDENTIFY EXACT DUPLICATES
    # Across ALL columns
    # ------------------------------------------------------------

    duplicate_mask = original_df.duplicated(
        keep=False
    )

    duplicate_records = original_df.loc[
        duplicate_mask
    ].copy()

    # Count rows that would be removed
    duplicate_rows_to_remove = int(
        original_df.duplicated(keep='first').sum()
    )

    # Number of rows participating in duplicate groups
    total_rows_in_duplicate_groups = len(
        duplicate_records
    )

    # Number of distinct duplicate patterns
    duplicate_patterns = (
        duplicate_records.value_counts(
            dropna=False
        ).rename('Occurrences').reset_index()
    )

    # Keep only actual repeated patterns
    duplicate_patterns = duplicate_patterns[
        duplicate_patterns['Occurrences'] > 1
    ]

    total_duplicate_groups = len(
        duplicate_patterns
    )

    # Largest duplicate group
    largest_duplicate_group = (
        int(duplicate_patterns['Occurrences'].max())
        if not duplicate_patterns.empty else 0
    )

    # ------------------------------------------------------------
    # C. COLUMN-WISE REPEATED VALUE ANALYSIS
    # Repeated values are NOT necessarily duplicates
    # ------------------------------------------------------------

    column_results = []

    for col in original_df.columns:

        value_counts = original_df[col].value_counts(
            dropna=True
        )

        repeated_values = value_counts[
            value_counts > 1
        ]

        repeated_occurrences = int(
            (value_counts - 1).clip(lower=0).sum()
        )

        most_common_value = (
            value_counts.index[0]
            if not value_counts.empty else np.nan
        )

        most_common_count = (
            int(value_counts.iloc[0])
            if not value_counts.empty else 0
        )

        column_results.append({

            'Column': col,

            'Data_Type': str(original_df[col].dtype),

            'Unique_Values': int(
                original_df[col].nunique(dropna=True)
            ),

            'Repeated_Distinct_Values': int(
                len(repeated_values)
            ),

            'Repeated_Occurrences': repeated_occurrences,

            'Most_Common_Value': most_common_value,

            'Most_Common_Frequency': most_common_count,

            'Missing_Values': int(
                original_df[col].isnull().sum()
            )

        })

    column_report = pd.DataFrame(
        column_results
    )

    # ------------------------------------------------------------
    # D. BUSINESS-KEY DUPLICATE INVESTIGATION
    # These are potential patterns, NOT automatic deletions
    # ------------------------------------------------------------

    business_key_results = []

    for key_name, columns in REVIEW_SUBSETS.items():

        available_columns = [
            col for col in columns
            if col in original_df.columns
        ]

        if not available_columns:
            continue

        # Ignore records with missing values in the
        # selected business-key columns
        key_data = original_df[
            available_columns
        ].dropna()

        repeated_key_rows = int(
            key_data.duplicated(
                subset=available_columns,
                keep='first'
            ).sum()
        )

        unique_key_count = int(
            key_data.drop_duplicates().shape[0]
        )

        business_key_results.append({

            'Key': key_name,

            'Columns': ', '.join(
                available_columns
            ),

            'Repeated_Key_Occurrences': repeated_key_rows,

            'Unique_Key_Combinations': unique_key_count,

            'Action': 'Review only - not removed'

        })

    business_key_report = pd.DataFrame(
        business_key_results
    )

    # ------------------------------------------------------------
    # E. DISPLAY SAMPLE DUPLICATE RECORDS
    # ------------------------------------------------------------

    sample_duplicates = duplicate_records.head(
        SAMPLE_SIZE
    )

    # ------------------------------------------------------------
    # F. REMOVE EXACT DUPLICATES
    # Preserve first occurrence
    # ------------------------------------------------------------

    if KEEP_FIRST_OCCURRENCE:

        cleaned_df = original_df.drop_duplicates(
            keep='first'
        ).copy()

    else:

        raise ValueError(
            "Only keep-first duplicate removal is enabled."
        )

    total_rows_after = len(cleaned_df)

    records_removed = (
        total_rows_before - total_rows_after
    )

    # ------------------------------------------------------------
    # G. VALIDATE DUPLICATE REMOVAL
    # ------------------------------------------------------------

    remaining_duplicates = int(
        cleaned_df.duplicated().sum()
    )

    validation_status = (
        remaining_duplicates == 0
    )

    # ------------------------------------------------------------
    # H. BEFORE AND AFTER SUMMARY
    # ------------------------------------------------------------

    summary_report = pd.DataFrame({

        'Metric': [

            'Total Rows Before',

            'Total Columns',

            'Rows Involved in Duplicate Groups',

            'Distinct Duplicate Patterns',

            'Duplicate Records Removed',

            'Total Rows After',

            'Remaining Exact Duplicates',

            'Duplicate Removal Validation'

        ],

        'Value': [

            total_rows_before,

            total_columns,

            total_rows_in_duplicate_groups,

            total_duplicate_groups,

            records_removed,

            total_rows_after,

            remaining_duplicates,

            'PASSED' if validation_status else 'FAILED'

        ]

    })

    # ------------------------------------------------------------
    # I. PRINT ALL REPORTS
    # ------------------------------------------------------------

    print("\n" + "=" * 70)

    print("       DUPLICATE DETECTION AND REMOVAL REPORT")

    print("=" * 70)

    print("\n1. EXACT DUPLICATE STATISTICS")

    print("-" * 70)

    print("Total Records:", total_rows_before)

    print(
        "Rows Involved in Duplicate Groups:",
        total_rows_in_duplicate_groups
    )

    print(
        "Distinct Duplicate Patterns:",
        total_duplicate_groups
    )

    print(
        "Records That Can Be Removed:",
        duplicate_rows_to_remove
    )

    print(
        "Largest Duplicate Group:",
        largest_duplicate_group
    )

    print("\n2. SAMPLE DUPLICATE RECORDS")

    print("-" * 70)

    if not sample_duplicates.empty:

        print(
            sample_duplicates.to_string(index=False)
        )

    else:

        print("No exact duplicate records found.")

    print("\n3. TOP DUPLICATE PATTERNS")

    print("-" * 70)

    if not duplicate_patterns.empty:

        print(
            duplicate_patterns.head(
                TOP_DUPLICATE_PATTERNS
            ).to_string(index=False)
        )

    else:

        print("No duplicate patterns found.")

    print("\n4. COLUMN-WISE REPETITION ANALYSIS")

    print("-" * 70)

    print(
        column_report.to_string(index=False)
    )

    print("\n5. BUSINESS-KEY DUPLICATE INVESTIGATION")

    print("-" * 70)

    print(
        business_key_report.to_string(index=False)
    )

    print("\n6. DUPLICATE REMOVAL SUMMARY")

    print("-" * 70)

    print(
        summary_report.to_string(index=False)
    )

    print("\n7. FINAL VALIDATION")

    print("-" * 70)

    if validation_status:

        print(
            "SUCCESS: All exact duplicate records removed."
        )

        print(
            "First occurrence of each record preserved."
        )

    else:

        print(
            "WARNING: Exact duplicates still exist."
        )

    print("\n" + "=" * 70)

    print("       DUPLICATE HANDLING COMPLETED")

    print("=" * 70)

    # ------------------------------------------------------------
    # J. RETURN ALL OUTPUTS
    # ------------------------------------------------------------

    return (

        cleaned_df,

        summary_report,

        duplicate_patterns,

        sample_duplicates,

        column_report,

        business_key_report

    )


# ================================================================
# SECTION 3: EXECUTE DUPLICATE HANDLING
# ================================================================

(
    deduplicated_df,
    duplicate_summary,
    duplicate_patterns,
    sample_duplicates,
    column_report,
    business_key_report

) = handle_duplicates(INPUT_DF)


# ================================================================
# SECTION 4: EXPORT ALL REPORTS TO EXCEL
# ================================================================

if EXPORT_REPORTS:

    with pd.ExcelWriter(
        EXPORT_FILE,
        engine='openpyxl'
    ) as writer:

        deduplicated_df.to_excel(
            writer,
            sheet_name='Deduplicated Dataset',
            index=False
        )

        duplicate_summary.to_excel(
            writer,
            sheet_name='Duplicate Summary',
            index=False
        )

        duplicate_patterns.to_excel(
            writer,
            sheet_name='Duplicate Patterns',
            index=False
        )

        sample_duplicates.to_excel(
            writer,
            sheet_name='Sample Duplicates',
            index=False
        )

        column_report.to_excel(
            writer,
            sheet_name='Column Analysis',
            index=False
        )

        business_key_report.to_excel(
            writer,
            sheet_name='Business Key Analysis',
            index=False
        )

    print("\nExcel report exported successfully:")
    print(EXPORT_FILE)


# ================================================================
# SECTION 5: FINAL DATASET VALIDATION
# ================================================================

print("\nFINAL DATASET VALIDATION")

print("-" * 70)

print("Original Dataset Shape:", INPUT_DF.shape)

print("Deduplicated Dataset Shape:", deduplicated_df.shape)

print(
    "Records Removed:",
    len(INPUT_DF) - len(deduplicated_df)
)

print(
    "Remaining Exact Duplicates:",
    deduplicated_df.duplicated().sum()
)

print("\nDeduplicated Dataset Preview:")

display(deduplicated_df.head())
```

# Prompt

Create a Python function for automatic data type inference and memory optimization of my cleaned Online Retail dataset. Safely optimize integer and float columns, convert suitable object columns to categories, preserve identifiers and financial precision, and keep InvoiceDate as datetime. Include before-and-after data type comparison, memory usage statistics, dataset validation, and export the optimized dataset to Excel."

```python

import pandas as pd
import numpy as np

# ============================================================
# 1. CONFIGURATION
# ============================================================

INPUT_DF = deduplicated_df

# Columns that should remain identifiers/text
IDENTIFIER_COLUMNS = ['InvoiceNo', 'StockCode']

# Columns that should be treated as categorical when beneficial
CATEGORICAL_COLUMNS = ['Description', 'Country']

# Financial column: preserve float64 precision
FINANCIAL_COLUMNS = ['UnitPrice']

# Output file
OUTPUT_FILE = 'online_retail_type_optimized.xlsx'


# ============================================================
# 2. AUTOMATIC DATA TYPE INFERENCE FUNCTION
# ============================================================

def automatic_type_inference(df):

    df_optimized = df.copy()
    conversion_stats = {}

    # --------------------------------------------------------
    # A. Optimize integer columns
    # --------------------------------------------------------

    for col in df_optimized.select_dtypes(
        include=['int64', 'int32']
    ).columns:

        col_min = df_optimized[col].min()
        col_max = df_optimized[col].max()

        if col_min >= 0:

            if col_max <= np.iinfo(np.uint8).max:
                new_type = 'uint8'

            elif col_max <= np.iinfo(np.uint16).max:
                new_type = 'uint16'

            elif col_max <= np.iinfo(np.uint32).max:
                new_type = 'uint32'

            else:
                new_type = 'uint64'

        else:

            if (
                col_min >= np.iinfo(np.int8).min
                and col_max <= np.iinfo(np.int8).max
            ):
                new_type = 'int8'

            elif (
                col_min >= np.iinfo(np.int16).min
                and col_max <= np.iinfo(np.int16).max
            ):
                new_type = 'int16'

            elif (
                col_min >= np.iinfo(np.int32).min
                and col_max <= np.iinfo(np.int32).max
            ):
                new_type = 'int32'

            else:
                new_type = 'int64'

        old_type = str(df_optimized[col].dtype)

        df_optimized[col] = df_optimized[col].astype(new_type)

        conversion_stats[col] = {
            'from': old_type,
            'to': str(df_optimized[col].dtype)
        }

    # --------------------------------------------------------
    # B. Handle CustomerID separately
    # --------------------------------------------------------

    if 'CustomerID' in df_optimized.columns:

        customer_values = pd.to_numeric(
            df_optimized['CustomerID'],
            errors='coerce'
        )

        non_missing = customer_values.dropna()

        if (
            (non_missing % 1 == 0).all()
            and (non_missing >= 0).all()
        ):

            old_type = str(df_optimized['CustomerID'].dtype)

            df_optimized['CustomerID'] = (
                customer_values.astype('Int64')
            )

            conversion_stats['CustomerID'] = {
                'from': old_type,
                'to': str(df_optimized['CustomerID'].dtype)
            }

    # --------------------------------------------------------
    # C. Optimize suitable floating-point columns
    # --------------------------------------------------------

    for col in df_optimized.select_dtypes(
        include=['float64']
    ).columns:

        if (
            col in FINANCIAL_COLUMNS
            or col == 'CustomerID'
        ):
            continue

        old_type = str(df_optimized[col].dtype)

        df_optimized[col] = pd.to_numeric(
            df_optimized[col],
            downcast='float'
        )

        conversion_stats[col] = {
            'from': old_type,
            'to': str(df_optimized[col].dtype)
        }

    # --------------------------------------------------------
    # D. Convert suitable object columns to category
    # --------------------------------------------------------

    for col in df_optimized.select_dtypes(
        include=['object', 'string']
    ).columns:

        if col in IDENTIFIER_COLUMNS:
            continue

        total_values = len(df_optimized[col])

        if total_values == 0:
            continue

        unique_values = df_optimized[col].nunique(
            dropna=True
        )

        unique_ratio = unique_values / total_values

        if unique_ratio < 0.5:

            old_type = str(df_optimized[col].dtype)

            df_optimized[col] = (
                df_optimized[col].astype('category')
            )

            conversion_stats[col] = {
                'from': old_type,
                'to': str(df_optimized[col].dtype)
            }

    # --------------------------------------------------------
    # E. Preserve InvoiceDate as datetime
    # --------------------------------------------------------

    if 'InvoiceDate' in df_optimized.columns:

        old_type = str(df_optimized['InvoiceDate'].dtype)

        df_optimized['InvoiceDate'] = pd.to_datetime(
            df_optimized['InvoiceDate'],
            errors='raise'
        )

        conversion_stats['InvoiceDate'] = {
            'from': old_type,
            'to': str(df_optimized['InvoiceDate'].dtype)
        }

    return df_optimized, conversion_stats


# ============================================================
# 3. APPLY AUTOMATIC TYPE INFERENCE
# ============================================================

df_type_optimized, type_stats = automatic_type_inference(
    INPUT_DF
)


# ============================================================
# 4. DISPLAY DATA TYPE CONVERSION REPORT
# ============================================================

print("=" * 60)
print("AUTOMATIC DATA TYPE CONVERSION REPORT")
print("=" * 60)

for column, stats in type_stats.items():

    print(
        f"{column}: "
        f"{stats['from']} -> {stats['to']}"
    )


# ============================================================
# 5. COMPARE DATA TYPES BEFORE AND AFTER OPTIMIZATION
# ============================================================

comparison = pd.DataFrame({

    'Before Optimization': INPUT_DF.dtypes.astype(str),

    'After Optimization': df_type_optimized.dtypes.astype(str)

})

print("\nDATA TYPE COMPARISON")

display(comparison)


# ============================================================
# 6. MEASURE MEMORY USAGE
# ============================================================

memory_before = (
    INPUT_DF.memory_usage(deep=True).sum()
    / (1024 ** 2)
)

memory_after = (
    df_type_optimized.memory_usage(deep=True).sum()
    / (1024 ** 2)
)

memory_saved = memory_before - memory_after

percentage_saved = (
    memory_saved / memory_before * 100
    if memory_before > 0 else 0
)

print("\nMEMORY OPTIMIZATION REPORT")
print("-" * 40)

print(f"Memory Before: {memory_before:.2f} MB")
print(f"Memory After:  {memory_after:.2f} MB")
print(f"Memory Saved:  {memory_saved:.2f} MB")
print(f"Memory Reduction: {percentage_saved:.2f}%")


# ============================================================
# 7. VALIDATE OPTIMIZED DATASET
# ============================================================

print("\nDATASET VALIDATION")
print("-" * 40)

print("Original Shape:", INPUT_DF.shape)
print("Optimized Shape:", df_type_optimized.shape)

print(
    "Missing Values Before:",
    INPUT_DF.isnull().sum().sum()
)

print(
    "Missing Values After:",
    df_type_optimized.isnull().sum().sum()
)

print(
    "Duplicate Rows Before:",
    INPUT_DF.duplicated().sum()
)

print(
    "Duplicate Rows After:",
    df_type_optimized.duplicated().sum()
)

# Check whether row count and column count are preserved
assert INPUT_DF.shape == df_type_optimized.shape

# Check that no new missing values were introduced
assert (
    df_type_optimized.isnull().sum().sum()
    <= INPUT_DF.isnull().sum().sum()
)

print("\nDataset structure validation passed!")


# ============================================================
# 8. DISPLAY OPTIMIZED DATASET
# ============================================================

print("\nOPTIMIZED DATASET PREVIEW")

display(df_type_optimized.head())


# ============================================================
# 9. SAVE OPTIMIZED DATASET
# ============================================================

df_type_optimized.to_excel(
    OUTPUT_FILE,
    index=False
)

print(f"\nOptimized dataset saved as: {OUTPUT_FILE}")
```
